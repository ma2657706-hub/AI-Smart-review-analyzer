```python
import gradio as gr
from transformers import pipeline

print("AI model load ho raha hai...")

analyzer = pipeline(
    "sentiment-analysis",
    model="distilbert/distilbert-base-uncased-finetuned-sst-2-english"
)

def analyze_review(review):
    if not review or not review.strip():
        return "Please enter a review.", ""

    result = analyzer(review.strip())[0]
    label = result["label"]
    confidence = round(result["score"] * 100, 2)

    if label == "POSITIVE":
        message = "😊 Positive Review"
    else:
        message = "☹️ Negative Review"

    return message, f"Confidence: {confidence}%"

demo = gr.Interface(
    fn=analyze_review,
    inputs=gr.Textbox(
        lines=4,
        label="Enter an English Review",
        placeholder="This product is amazing!"
    ),
    outputs=[
        gr.Textbox(label="Sentiment Result"),
        gr.Textbox(label="Model Confidence")
    ],
    title="AI Smart Review Analyzer",
    description="Analyze English reviews using an AI sentiment model.",
    examples=[
        ["This product is amazing. I love it!"],
        ["Very poor quality. I am disappointed."],
        ["The product is good and works well."]
    ]
)

demo.launch(share=True)
```

```python
%%writefile app.py
import gradio as gr
from transformers import pipeline

analyzer = pipeline(
    "sentiment-analysis",
    model="distilbert/distilbert-base-uncased-finetuned-sst-2-english"
)

def analyze_review(review):
    if not review or not review.strip():
        return "Please enter a review.", "No result"

    result = analyzer(review.strip())[0]
    label = result["label"]
    confidence = round(result["score"] * 100, 2)

    sentiment = "😊 Positive Review" if label == "POSITIVE" else "☹️ Negative Review"
    return sentiment, f"{confidence}% confidence"

demo = gr.Interface(
    fn=analyze_review,
    inputs=gr.Textbox(lines=4, label="Enter an English Review"),
    outputs=[
        gr.Textbox(label="Sentiment"),
        gr.Textbox(label="Confidence")
    ],
    title="AI Smart Review Analyzer",
    description="Analyze English product reviews using AI.",
    examples=[
        ["This product is amazing. I love it!"],
        ["Very poor quality. I am disappointed."]
    ]
)

demo.launch()
```
